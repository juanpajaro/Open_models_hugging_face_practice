```python
from sentence_transformers import SentenceTransformer, util
```

```python
model = SentenceTransformer("./model_similarity")
```

```python
sentences = [
    "The weather is lovely today.",
    "It's so sunny outside!",
    "He drove to the stadium."
]
```

```python
print(len(sentences))
```

```python
print("Embeddings:")
for sentence, embedding in zip(sentences, embeddings):
    print(f"Sentence: {sentence}")
    print(f"Embedding: {embedding[:5]}...")  # Print first 5 dimensions for brevity
    print()
```

## ¿que es un valor promedio?
Un valor promedio es un número representativo que resume un conjunto de datos o muestra su punto central. En el uso cotidiano, suele referirse a la media aritmética, que se obtiene sumando todos los valores y dividiendo el resultado entre la cantidad total de datos

```python
correct_answer = ["Un valor promedio es un número representativo que resume un conjunto de datos o muestra su punto central. En el uso cotidiano, suele referirse a la media aritmética, que se obtiene sumando todos los valores y dividiendo el resultado entre la cantidad total de datos."]
```

```python
emb_answer = model.encode(correct_answer, convert_to_tensor=True)
print(emb_answer.shape)
```

```python
student_answer = ["es el valor central dentro de un conjunto de datos."]
```

```python
emb_student = model.encode(student_answer, convert_to_tensor=True)
print(emb_student.shape)
```

```python
cosine_scores = util.cos_sim(emb_student,emb_answer)
print(cosine_scores)
```
