````python
from transformers import pipeline
````

````python
pipe = pipeline("text-generation", model="Qwen/Qwen3-0.6B")
messages = [
    {"role": "user", "content": "Who are you?"},
]
pipe(messages)
````

````python
messages = [{"role": "user", "content": "Quien es el presidente de Venezuela?"},]
pipe(messages)
````

````python
messages = [{"role": "user", "content": "Como hago para ser feliz"},]
pipe(messages)
````

````python
pipe_local = pipeline(
    "text-generation",
    model="/home/pajaro/Open_models_hugging_face_practice/model",
)
````

````python
messages = [{"role": "user", "content": "Quien es el presidente de Venezuela?"},]
result = pipe_local(messages, max_new_tokens=256)
print(result[0]["generated_text"][-1]["content"])
````

````python
messages = [{"role": "user", "content": "Quien es el presidente de Venezuela?"},]
result = pipe_local(messages, max_new_tokens=256, tokenizer_encode_kwargs={"enable_thinking": False},)
print(result[0]["generated_text"][-1]["content"])
````

````python
messages = [{"role": "user", "content": "¿Que recomendaciones tienes para aprender a tocar la guitarra?"},]
result = pipe_local(messages, max_new_tokens=256, tokenizer_encode_kwargs={"enable_thinking": False},)
print(result[0]["generated_text"][-1]["content"])
````
